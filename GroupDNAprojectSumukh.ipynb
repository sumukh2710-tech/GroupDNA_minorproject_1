```python
# GroupDNA - Your WhatsApp Group Chat, Decoded
file_path = "chat.txt"
with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()
print("Total lines:", len(lines))
```

```python
for line in lines[:10]:
    print(line.strip())
```

```python
messages = []
current_message = None

for line in lines:

    line = line.rstrip("\n")

    # Check whether the line starts a new WhatsApp message
    if line.startswith("[") and "] " in line:

        header, text = line.split("] ", 1)

        # Extract timestamp
        timestamp_text = header[1:]

        # Check whether this is a real user message
        if ": " in text:
            sender, message_text = text.split(": ", 1)

            # Ignore system messages
            if not sender.startswith("-"):

                current_message = {
                    "timestamp": timestamp_text,
                    "sender": sender,
                    "text": message_text
                }

                messages.append(current_message)

            else:
                current_message = None

        else:
            current_message = None

    else:
        # Multiline message
        if current_message is not None:
            current_message["text"] += " " + line.strip()

print("Parsed messages:", len(messages))
```

```python
media_markers = [
    "<image omitted>",
    "<video omitted>",
    "<sticker omitted>",
    "<voice message omitted>",
    "<document omitted>",
    "<album message>",
    "<unknown message>",
    "<audio omitted>",
    "<contact omitted>",
    "<location omitted>"
]

clean_messages = []

for message in messages:

    text = message["text"].strip()

    # Deleted messages
    if text == "This message was deleted":
        continue

    if text == "You deleted this message":
        continue

    # Media-only messages
    if text in media_markers:
        continue

    # Remove media markers if text exists along with them
    for marker in media_markers:
        text = text.replace(marker, "").strip()

    if text == "":
        continue

    clean_messages.append({
        "timestamp": message["timestamp"],
        "sender": message["sender"],
        "text": text
    })

print("Clean messages:", len(clean_messages))
```

```python
punctuation = ".,!?;:()[]{}\"'/-_@#$%^&*+=<>|\\"

clean_text = ""

for char in all_text:

    if char in punctuation:
        clean_text += " "
    else:
        clean_text += char

print(clean_text[:500])
```

```python
words = clean_text.split()

print("Total words:", len(words))
print(words[:50])
```

```python
stop_words = {
    "the", "is", "a", "an", "and", "or", "to", "of",
    "in", "on", "for", "with", "this", "that", "it",
    "was", "are", "be", "i", "you", "we", "he", "she",
    "they", "me", "my", "your", "our", "but", "so",
    "if", "at", "from", "as", "have", "has", "had",
    "will", "can", "do", "did", "not"
}
```

```python
useful_words = []

for word in words:

    if word not in stop_words and len(word) > 1:
        useful_words.append(word)

print("Useful words:", len(useful_words))
print(useful_words[:50])
```

```python
word_frequency = {}

for word in useful_words:

    if word not in word_frequency:
        word_frequency[word] = 0

    word_frequency[word] += 1
```

```python
for word, count in list(word_frequency.items())[:20]:
    print(word, ":", count)
```

```python
frequency_copy = word_frequency.copy()

top_words = []

for i in range(20):

    if len(frequency_copy) == 0:
        break

    most_common_word = None
    highest_frequency = 0

    for word, count in frequency_copy.items():

        if count > highest_frequency:
            highest_frequency = count
            most_common_word = word

    top_words.append((most_common_word, highest_frequency))

    del frequency_copy[most_common_word]
```

```python
person_words = {}

for person in participants:
    person_words[person] = {}
```

```python
for message in clean_messages:

    person = message["sender"]
    text = message["text"].lower()

    current_word = ""

    for char in text:

        if char.isalnum():
            current_word += char

        else:
            if current_word != "" and current_word not in stop_words:

                if current_word not in person_words[person]:
                    person_words[person][current_word] = 0

                person_words[person][current_word] += 1

                current_word = ""

    # Handle final word
    if current_word != "" and current_word not in stop_words:

        if current_word not in person_words[person]:
            person_words[person][current_word] = 0

        person_words[person][current_word] += 1
```

```python
print("FAVORITE WORD BY PARTICIPANT")
print("=" * 40)

for person in person_words:

    highest_word = None
    highest_count = 0

    for word, count in person_words[person].items():

        if count > highest_count:
            highest_word = word
            highest_count = count

    print(
        person,
        "->",
        highest_word,
        "(",
        highest_count,
        "times)"
    )
```

```python
emoji_frequency = {}

for message in clean_messages:

    for char in message["text"]:

        if not char.isalnum() and not char.isspace():

            if char not in punctuation:

                if char not in emoji_frequency:
                    emoji_frequency[char] = 0

                emoji_frequency[char] += 1
```

```python
emoji_copy = emoji_frequency.copy()

top_emojis = []

for i in range(10):

    if len(emoji_copy) == 0:
        break

    highest_emoji = None
    highest_count = 0

    for emoji, count in emoji_copy.items():

        if count > highest_count:
            highest_emoji = emoji
            highest_count = count

    top_emojis.append((highest_emoji, highest_count))

    del emoji_copy[highest_emoji]
```

```python
response_times = []

for i in range(1, len(clean_messages)):

    previous_message = clean_messages[i - 1]
    current_message = clean_messages[i]

    previous_sender = previous_message["sender"]
    current_sender = current_message["sender"]

    # Only calculate when another person responds
    if previous_sender != current_sender:

        time_difference = (
            current_message["datetime"]
            - previous_message["datetime"]
        )

        seconds = time_difference.total_seconds()

        response_times.append({
            "from": previous_sender,
            "to": current_sender,
            "seconds": seconds,
            "minutes": seconds / 60
        })

print("Response interactions:", len(response_times))
```

```python
for response in response_times[:20]:

    print(
        response["from"],
        "→",
        response["to"],
        ":",
        round(response["minutes"], 2),
        "minutes"
    )
```

```python
fastest_response = None

for response in response_times:

    if response["seconds"] <= 0:
        continue

    if fastest_response is None:
        fastest_response = response

    elif response["seconds"] < fastest_response["seconds"]:
        fastest_response = response


print("FASTEST RESPONSE")
print("-" * 30)

print(
    fastest_response["from"],
    "→",
    fastest_response["to"]
)

print(
    "Response time:",
    round(fastest_response["minutes"], 2),
    "minutes"
)
```

```python
response_window = 30 * 60

valid_responses = []

for response in response_times:

    if response["seconds"] <= response_window:

        valid_responses.append(response)

print("Valid responses within 30 minutes:",
      len(valid_responses))
```

```python
total_response_seconds = 0

for response in valid_responses:
    total_response_seconds += response["seconds"]

if len(valid_responses) > 0:

    average_response_seconds = (
        total_response_seconds / len(valid_responses)
    )

    average_response_minutes = (
        average_response_seconds / 60
    )

    print(
        "Average response time:",
        round(average_response_minutes, 2),
        "minutes"
    )

else:
    print("No valid responses found.")
```

```python
participant_response = {}

for response in valid_responses:

    person = response["to"]

    if person not in participant_response:

        participant_response[person] = {
            "total_seconds": 0,
            "count": 0
        }

    participant_response[person]["total_seconds"] += response["seconds"]
    participant_response[person]["count"] += 1
```

```python
print("AVERAGE RESPONSE TIME BY PARTICIPANT")
print("=" * 45)

for person, data in participant_response.items():

    average_seconds = (
        data["total_seconds"] / data["count"]
    )

    average_minutes = average_seconds / 60

    print(
        person,
        ":",
        round(average_minutes, 2),
        "minutes",
        "(",
        data["count"],
        "responses)"
    )
```

```python
silence_threshold = 6 * 60 * 60

silence_periods = []

for i in range(1, len(clean_messages)):

    previous_message = clean_messages[i - 1]
    current_message = clean_messages[i]

    gap = (
        current_message["datetime"]
        - previous_message["datetime"]
    )

    gap_seconds = gap.total_seconds()

    if gap_seconds >= silence_threshold:

        silence_periods.append({
            "start": previous_message["datetime"],
            "end": current_message["datetime"],
            "seconds": gap_seconds,
            "hours": gap_seconds / 3600
        })
```

```python
print("LONG SILENCE PERIODS")
print("=" * 45)

for silence in silence_periods:

    print(
        silence["start"],
        "→",
        silence["end"]
    )

    print(
        "Silence:",
        round(silence["hours"], 2),
        "hours"
    )

    print("-" * 45)
```

```python
longest_silence = None

for silence in silence_periods:

    if longest_silence is None:
        longest_silence = silence

    elif silence["seconds"] > longest_silence["seconds"]:
        longest_silence = silence


if longest_silence is not None:

    print("LONGEST SILENCE")
    print("=" * 30)

    print("Start :", longest_silence["start"])
    print("End   :", longest_silence["end"])
    print(
        "Duration:",
        round(longest_silence["hours"], 2),
        "hours"
    )

else:

    print("No silence period above 6 hours found.")
```

```python
conversation_bursts = []

current_burst = []

for i in range(len(clean_messages)):

    if len(current_burst) == 0:

        current_burst.append(clean_messages[i])

    else:

        previous = current_burst[-1]
        current = clean_messages[i]

        gap = (
            current["datetime"]
            - previous["datetime"]
        ).total_seconds()

        if gap <= 30 * 60:

            current_burst.append(current)

        else:

            if len(current_burst) > 1:
                conversation_bursts.append(current_burst)

            current_burst = [current]


if len(current_burst) > 1:
    conversation_bursts.append(current_burst)
```
